# Task 2: Model-Based Agent - Intelligent Patient Monitoring

```python

class PatientMonitorAgent:
    def __init__(self):
        self.memory = [] 
    def update_memory(self, reading):
      
        self.memory.append(reading)

    def decide(self, reading):
        self.update_memory(reading)

        hr, bp_sys, bp_dia, spo2 = reading["HR"], reading["BP_sys"], reading["BP_dia"], reading["SpO2"]

        
        if len(self.memory) > 1:
            prev = self.memory[-2]
            hr_rising = hr > prev["HR"] + 5
            bp_rising = bp_sys > prev["BP_sys"] + 5
        else:
            hr_rising = bp_rising = False

    
        if spo2 < 90 or hr > 130 or bp_sys > 180:
            return "Call Doctor"
      
        elif spo2 < 95 or hr > 100 or bp_sys > 140 or hr_rising or bp_rising:
            return "Alert Nurse"
        else:
            return "Continue Monitoring"


readings = [
    {"HR": 98, "BP_sys": 130, "BP_dia": 85, "SpO2": 97},
    {"HR": 105, "BP_sys": 140, "BP_dia": 90, "SpO2": 93},
    {"HR": 112, "BP_sys": 150, "BP_dia": 95, "SpO2": 92},
    {"HR": 135, "BP_sys": 185, "BP_dia": 100, "SpO2": 88},
]

agent = PatientMonitorAgent()
for i, r in enumerate(readings, start=1):
    action = agent.decide(r)
    print(f"Reading {i}: {r} -> Action: {action}")
```

# Task 3: Goal-Based Agent - Autonomous Warehouse Logistics

```python

class WarehouseRobotAgent:
    def __init__(self, goal):
        self.goal = goal  

    def goal_test(self, state):
        
        return state == self.goal

    def act(self, state):
       
        x, y = state
        gx, gy = self.goal
        if x < gx:
            x += 1
        elif x > gx:
            x -= 1
        elif y < gy:
            y += 1
        elif y > gy:
            y -= 1
        return (x, y)

    def run(self, start_state):
        state = start_state
        path = [state]
        while not self.goal_test(state):
            state = self.act(state)
            path.append(state)
        return path


goal = (5, 5)  
start_positions = [(0, 0), (2, 4), (5, 1)]

agent = WarehouseRobotAgent(goal)
for start in start_positions:
    path = agent.run(start)
    print(f"Start: {start} -> Path: {path} -> Goal Reached: {agent.goal_test(path[-1])}")
```
