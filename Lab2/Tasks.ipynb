# PHASE 4: INVESTIGATE OUTLIERS

### TASK 7: Statistical Outlier Identification Using IQR

```python
def iqr_analysis(df, column):
    """Calculate IQR bounds and count outliers for a given column."""
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(f'--- {column} ---')
    print(f'  Q1          : {Q1}')
    print(f'  Q3          : {Q3}')
    print(f'  IQR         : {IQR}')
    print(f'  Lower Bound : {lower_bound}')
    print(f'  Upper Bound : {upper_bound}')
    print(f'  Outliers    : {len(outliers)}')
    print()
    return Q1, Q3, IQR, lower_bound, upper_bound
```

```python
# SalePrice
saleprice_bounds = iqr_analysis(
    house_df,
    "SalePrice"
)
```

```python
# GrLivArea
gl_bounds = iqr_analysis(
    house_df,
    "GrLivArea"
)
```

```python
# GarageArea
garage_bounds = iqr_analysis(
    house_df,
    "GarageArea"
)
```

```python
# Box Plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(y=house_df['SalePrice'], ax=axes[0], color='skyblue')
axes[0].set_title('SalePrice Box Plot')
axes[0].set_ylabel('SalePrice ($)')

sns.boxplot(y=house_df['GrLivArea'], ax=axes[1], color='salmon')
axes[1].set_title('GrLivArea Box Plot')
axes[1].set_ylabel('GrLivArea (sq ft)')

sns.boxplot(y=house_df['GarageArea'], ax=axes[2], color='lightgreen')
axes[2].set_title('GarageArea Box Plot')
axes[2].set_ylabel('GarageArea (sq ft)')

plt.tight_layout()
plt.show()
```

### TASK 8: Outlier Mitigation Through IQR-Based Filtering

```python
# Using SalePrice and GrLivArea IQR bounds from Task 7 to filter
sp_Q1, sp_Q3, sp_IQR, sp_lower, sp_upper = saleprice_bounds
gl_Q1, gl_Q3, gl_IQR, gl_lower, gl_upper = gl_bounds

# Create cleaned dataset by removing observations outside IQR bounds
clean_house_df = house_df[
    (house_df['SalePrice'] >= sp_lower) & (house_df['SalePrice'] <= sp_upper) &
    (house_df['GrLivArea'] >= gl_lower) & (house_df['GrLivArea'] <= gl_upper)
].copy()
```

```python
print("Original shape:", house_df.shape)
print("Cleaned shape :", clean_house_df.shape)
```

```python
print("SalePrice:", sp_lower, sp_upper)
print("GrLivArea:", gl_lower, gl_upper)

print("\nActual data:")
print(house_df[['SalePrice', 'GrLivArea']].describe())
```

```python
# Percentage of records removed
# Percentage of observations removed
removed = house_df.shape[0] - clean_house_df.shape[0]
pct_removed = (removed / house_df.shape[0]) * 100
print(f'Observations removed: {removed}')
print(f'Percentage removed  : {pct_removed:.2f}%')
```
