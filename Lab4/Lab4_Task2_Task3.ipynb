# Task 2

```python
from collections import deque

actions = {
    "S0": [("Allocate Resources", "S1", 2)],
    "S1": [("Create VM", "S2", 4)],
    "S2": [("Install Dependencies", "S3", 3)],
    "S3": [("Deploy AI Model", "S4", 5)],
    "S4": [("Verify Deployment", "S5", 2)],
    "S5": [("Confirm Deployment", "S6", 1)],
}

initial_state = "S0"
goal_state = "S6"

visited = set()
queue = deque([initial_state])
while queue:
    state = queue.popleft()
    if state in visited:
        continue
    visited.add(state)
    for _, next_state, _ in actions.get(state, []):
        queue.append(next_state)
print("State space:", sorted(visited))

queue = deque([(initial_state, [], 0)])
visited = set()
while queue:
    state, path, cost = queue.popleft()
    if state == goal_state:
        break
    if state in visited:
        continue
    visited.add(state)
    for action, next_state, step_cost in actions.get(state, []):
        queue.append((next_state, path + [(state, action, next_state, step_cost)], cost + step_cost))

for state, action, next_state, step_cost in path:
    print(f"{state} --[{action} ({step_cost} min)]--> {next_state}")

sequence = [initial_state] + [step[2] for step in path]
print("\nExecution sequence:", " -> ".join(sequence))
print("Total deployment cost:", cost, "min")
```

# Task 3

```python
from collections import deque

actions = {
    "S0": [("Authenticate Student", "S1", 1)],
    "S1": [("Verify Prerequisites", "S2", 2)],
    "S2": [("Select Courses", "S3", 2)],
    "S3": [("Verify Fee Status", "S4", 1)],
    "S4": [("Confirm Enrollment", "S5", 1)],
}

initial_state = "S0"
goal_state = "S5"

visited = set()
queue = deque([initial_state])
while queue:
    state = queue.popleft()
    if state in visited:
        continue
    visited.add(state)
    for _, next_state, _ in actions.get(state, []):
        queue.append(next_state)
print("Reachable states:", sorted(visited))

queue = deque([(initial_state, [], 0)])
visited = set()
while queue:
    state, path, cost = queue.popleft()
    if state == goal_state:
        break
    if state in visited:
        continue
    visited.add(state)
    for action, next_state, step_cost in actions.get(state, []):
        if next_state not in visited:
            queue.append((next_state, path + [(state, action, next_state, step_cost)], cost + step_cost))

for state, action, next_state, step_cost in path:
    print(f"{state} --[{action} (cost {step_cost})]--> {next_state}")

sequence = [initial_state] + [step[2] for step in path]
print("\nExecution sequence:", " -> ".join(sequence))
print("Total path cost:", cost)
```
